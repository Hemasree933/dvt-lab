```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('employee_salaries.csv')

# 1. Line Plot: Individual Employee Salaries
plt.figure(figsize=(8, 4))
plt.plot(df['Name'], df['Salary'], marker='o', color='royalblue', linewidth=2, markersize=8)
plt.title('Employee Salary Trend', fontsize=12, fontweight='bold')
plt.xlabel('Employee Name', fontsize=10)
plt.ylabel('Salary (INR)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# 2. Bar Plot: Average Salary by Department
dept_avg = df.groupby('Department')['Salary'].mean()

plt.figure(figsize=(8, 4))
plt.bar(dept_avg.index, dept_avg.values, color='skyblue', edgecolor='black', width=0.5)
plt.title('Average Salary per Department', fontsize=12, fontweight='bold')
plt.xlabel('Department', fontsize=10)
plt.ylabel('Mean Salary (INR)', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```

```python
# 1. Histogram: Distribution of Experience
plt.figure(figsize=(7, 4))
plt.hist(df['Experience_Years'], bins=5, color='mediumseagreen', edgecolor='black', alpha=0.7)
plt.title('Distribution of Employee Experience', fontsize=12, fontweight='bold')
plt.xlabel('Experience (Years)', fontsize=10)
plt.ylabel('Frequency', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()

# 2. Scatter Plot: Experience vs. Salary
plt.figure(figsize=(7, 4))
plt.scatter(df['Experience_Years'], df['Salary'], color='darkorchid', s=100, edgecolors='black')
plt.title('Salary vs Experience', fontsize=12, fontweight='bold')
plt.xlabel('Years of Experience', fontsize=10)
plt.ylabel('Salary (INR)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# 3. Box Plot: Salary Distribution & Outlier Identification
plt.figure(figsize=(5, 4))
plt.boxplot(df['Salary'], patch_artist=True, boxprops=dict(facecolor='lightcoral'))
plt.title('Salary Distribution Spread', fontsize=12, fontweight='bold')
plt.ylabel('Salary (INR)', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```
