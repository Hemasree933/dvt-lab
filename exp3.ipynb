```python
import pandas as pd
import numpy as np

# Load original dataset
df = pd.read_csv('employee_salaries.csv')

# Ensure 'City' column exists if missing from your initial file
if 'City' not in df.columns:
    df['City'] = ['Chennai', 'Bangalore', 'Chennai', 'Hyderabad', 'Bangalore', 'Chennai', 'Coimbatore', 'Hyderabad'][:len(df)]

# Create the new rows to introduce duplicates and missing values
new_data = pd.DataFrame([
    [102, 'Bala', 'IT', 30000.0, 3.0, 'Bangalore'],
    [109, 'Suresh', np.nan, 39000.0, np.nan, 'Chennai'],
    [110, 'Kavya', 'HR', np.nan, 4.0, np.nan]
], columns=['EmpID', 'Name', 'Department', 'Salary', 'Experience_Years', 'City'])

# Combine datasets using concat
df = pd.concat([df, new_data], ignore_index=True)

print("==========================================")
print(" 1. RAW DATASET WITH MISSING & DUPLICATES ")
print("==========================================")
display(df)
```

```python
# 1. Check and remove duplicate rows
print("==========================================")
print(" 1. DUPLICATE REMOVAL                     ")
print("==========================================")
print("Duplicate rows count before:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("Duplicate rows count after:", df.duplicated().sum())

# 2. Identify missing values
print("\n==========================================")
print(" 2. MISSING VALUE DETECTION               ")
print("==========================================")
print(df.isnull().sum())

# 3. Handle missing values (Imputation)
# Fill numerical missing values with Mean
salary_mean = df['Salary'].mean()
exp_mean = df['Experience_Years'].mean()

df['Salary'].fillna(salary_mean, inplace=True)
df['Experience_Years'].fillna(exp_mean, inplace=True)

# Fill categorical missing values with Mode (Most Frequent)
dept_mode = df['Department'].mode()[0]
city_mode = df['City'].mode()[0]

df['Department'].fillna(dept_mode, inplace=True)
df['City'].fillna(city_mode, inplace=True)

print("\n==========================================")
print(" 3. CLEANED DATASET AFTER IMPUTATION     ")
print("==========================================")
display(df)
```

```python
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

# 1. Outlier Detection using IQR Method
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]
print("==========================================")
print(" 1. OUTLIER DETECTION (IQR METHOD)        ")
print("==========================================")
print(f"Salary Lower Bound: {lower_bound:.2f}")
print(f"Salary Upper Bound: {upper_bound:.2f}")
print("Number of detected Salary outliers:", len(outliers))

# 2. Feature Normalization (Min-Max Scaling)
scaler = MinMaxScaler()
df[['Salary_Scaled', 'Experience_Scaled']] = scaler.fit_transform(df[['Salary', 'Experience_Years']])

# 3. Categorical Label Encoding
encoder = LabelEncoder()
df['Department_Encoded'] = encoder.fit_transform(df['Department'])
df['City_Encoded'] = encoder.fit_transform(df['City'])

print("\n==========================================")
print(" 2. TRANSFORMED DATASET (ENCODED & SCALED)")
print("==========================================")
display(df[['EmpID', 'Name', 'Department', 'Department_Encoded', 'City', 'City_Encoded', 'Salary_Scaled', 'Experience_Scaled']])
```
