```python
import pandas as pd

# Creating a sample dataset for Experiment 2
data = {
    'EmpID': [101, 102, 103, 104, 105, 106],
    'Name': ['Arun', 'Bala', 'Divya', 'Priya', 'Ravi', 'Karthik'],
    'Department': ['HR', 'IT', 'CSE', 'AI&DS', 'IT', 'AI&DS'],
    'Salary': [25000, 30000, 45000, 38000, 32000, 41000],
    'Experience_Years': [2, 3, 7, 5, 4, 6]
}

# Save as CSV file
df_initial = pd.DataFrame(data)
df_initial.to_csv('employee_salaries.csv', index=False)
print("Dataset 'employee_salaries.csv' successfully created!")
```

```python
import pandas as pd

# Dataset Name: employee_salaries.csv
dataset_name = 'employee_salaries.csv'

# Program 1: Reading CSV File
df = pd.read_csv(dataset_name)

# Program 2: Display First Five Records
print("--- First 5 Records ---")
display(df.head())

# Program 3: Display Dataset Information
print("\n--- Dataset Information ---")
df.info()

# Program 4: Statistical Summary
print("\n--- Statistical Summary ---")
display(df.describe())
```

```python
# Program 8: Export to CSV
df.to_csv('output_employee.csv', index=False)
print("CSV File Exported successfully: output_employee.csv")

# Program 9: Export to Excel
df.to_excel('output_employee.xlsx', index=False)
print("Excel File Exported successfully: output_employee.xlsx")

# Program 10: Export to JSON
df.to_json('output_employee.json')
print("JSON File Exported successfully: output_employee.json")
```
