```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('employee_salaries.csv')

# Ensure 'City' column exists
if 'City' not in df.columns:
    df['City'] = ['Chennai', 'Bangalore', 'Chennai', 'Hyderabad', 'Bangalore', 'Chennai', 'Coimbatore', 'Hyderabad'][:len(df)]

# Power Query Pivot equivalent: Group by Department and calculate Average Salary
pivot_table = df.pivot_table(index='Department', values='Salary', aggfunc='mean').reset_index()

# Sort values descending (Equivalent to Power BI Sort Axis)
pivot_table = pivot_table.sort_values(by='Salary', ascending=False)

print("==========================================")
print(" 1. POWER BI PIVOT TABLE EQUIVALENT      ")
print("==========================================")
display(pivot_table)
```
