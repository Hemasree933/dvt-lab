```python
import pandas as pd

# Load the dataset
df = pd.read_csv('employee_salaries.csv')

print("==========================================")
print(" 1. FILTERING DATA (Salary > 35,000)      ")
print("==========================================")
high_salary_df = df[df['Salary'] > 35000]
display(high_salary_df)

print("\n==========================================")
print(" 2. MULTI-CONDITION FILTERING             ")
print(" (Salary >= 35000 AND Experience >= 5)    ")
print("==========================================")
filtered_df = df[(df['Salary'] >= 35000) & (df['Experience_Years'] >= 5)]
display(filtered_df)

print("\n==========================================")
print(" 3. SORTING DATA BY EXPERIENCE (DESCENDING)")
print("==========================================")
sorted_df = df.sort_values(by='Experience_Years', ascending=False)
display(sorted_df)
```

```python
# 1. Grouping by Department and calculating mean Salary & Experience
print("==========================================")
print(" 1. AVERAGE SALARY & EXPERIENCE BY DEPT   ")
print("==========================================")
dept_summary = df.groupby('Department')[['Salary', 'Experience_Years']].mean().reset_index()
display(dept_summary)

# 2. Advanced Aggregation (Count, Min, Max, Mean Salary by Department)
print("\n==========================================")
print(" 2. ADVANCED SALARY STATS BY DEPARTMENT   ")
print("==========================================")
dept_advanced = df.groupby('Department')['Salary'].agg(['count', 'min', 'max', 'mean']).reset_index()
display(dept_advanced)

# 3. Grouping by City
print("\n==========================================")
print(" 3. TOTAL SALARY EXPENSE BY CITY          ")
print("==========================================")
city_summary = df.groupby('City')['Salary'].sum().reset_index()
display(city_summary)
```
